```python
import numpy as np
import k3d

# 1. Create data (similar to meshgrid)
x = np.arange(-10, 10, 1)
y = np.arange(-10, 10, 1)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))

# 2. Create a structured grid
# grid = pv.StructuredGrid(x, y, z)

# 3. Plot with high-quality settings
plot = k3d.plot()
surface = k3d.surface(z, xmin=-10, xmax=10, ymin=-10, ymax=10, color_map=k3d.colormaps.basic_color_maps.Jet)
plot += surface
plot.display()
```
